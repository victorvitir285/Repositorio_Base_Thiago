```python
# criar um banco de dados com SQLite

import sqlite3

conexao = sqlite3.connect('dados.db')

cursor = conexao.cursor()
```

```python
cursor.execute('''

                    CREATE TABLE IF NOT EXISTS produtos (
                        id INTEGER PRIMARY KEY AUTOINCREMENT,
                        marca VARCHAR(50) NOT NULL,
                        produto VARCHAR(50) NOT NULL,
                        quantidade INTEGER,
                        valor DECIMAL NOT NULL      
                    ); 

        
                ''')

conexao.commit()
conexao.close()


```

```python
cursor.execute("INSERT INTO produtos ('marca','produtor','quantidade','valor') VALUES ('Moura','Bateria 60ah',10 , 500);")              
cursor.execute("INSERT INTO produtos ('marca','produtor','quantidade','valor') VALUES ('AC Delco','Bateria 60ah',15 , 450);")

conexao.commit()
conexao.close()
```

```python
conexao = sqlite3.connect('dados.db')
cursor = conexao.cursor()

consulta = cursor.execute("select * from produtos;")

imprimir = consulta.fetchall()

for i in imprimir:
    print(i)
```
