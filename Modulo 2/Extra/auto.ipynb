```python
musica = '''As andorinhas voltaram
E eu também voltei
Pousar no velho ninho
Que um dia aqui deixei
Nós somos andorinhas
Que vão e quem vem
À procura de amor
Às vezes volta cansada
Ferida, machucada
Mas volta pra casa
Batendo suas asas
Com grande dor
Igual à andorinha
Eu parti sonhando
Mas foi tudo em vão
Voltei sem felicidade
Porque na verdade
Uma andorinha voando sozinha
Não faz verão
Com vocês, Trio Parada Dura'''

print(musica)
```

```python
textoConvertido = musica.split('\n')

pyautogui.hotkey('winleft','r')
pyautogui.write('notepad')
pyautogui.press('enter')
time.sleep(2)

for i in textoConvertido:
    pyautogui.write(i, interval=0.4)
    pyautogui.press('enter')
```
